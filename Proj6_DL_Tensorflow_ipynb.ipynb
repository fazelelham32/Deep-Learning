#Data set

```python
from sklearn.datasets import make_classification

x, y = make_classification(n_samples=1000, n_features=20, n_informative=10, n_redundant=5, n_classes=2, random_state=0)
```

```python
x[0]
```

```python
y[0]
```

```python
x.shape, y.shape
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2)
```

```python
x_train.shape, y_train.shape
```

```python
x_test.shape, y_test.shape
```

#Model definition

```python
from tensorflow import keras
import tensorflow as tf
def create_model(hidden_units=16, optimizer='adam'):
  model = keras.Sequential([
  keras.layers.Dense(hidden_units, input_shape=(20,), activation='relu'),
  keras.layers.Dense(1, activation='sigmoid')
])
  model.compile(optimizer=optimizer, loss='binary_crossentropy', metrics=['accuracy'])

  return model
```

#Grid search

```python
from sklearn.model_selection import GridSearchCV
```

```python
!pip install -U scikeras scikit-learn
```

```python
from scikeras.wrappers import KerasClassifier
model = KerasClassifier(model=create_model, verbose=0)
```

```python
params = {
    "model__hidden_units": [16, 32, 64],
    "model__optimizer": ["adam", "rmsprop"],

}
```

```python
grid = GridSearchCV(estimator=model, param_grid=params, cv=3)
grid_result = grid.fit(x_train, y_train)
```

```python
best_params = grid_result.best_params_
best_params
```

```python
best_score = grid_result.best_score_
best_score
```

#Random search

```python
from sklearn.model_selection import RandomizedSearchCV

rs = RandomizedSearchCV(estimator=model, param_distributions=params, cv=3, n_iter=5)
rs_result = rs.fit(x_train, y_train)
```

```python
rs_best_params = rs_result.best_params_
rs_best_params
```

```python
rs_best_score = rs_result.best_score_
rs_best_score
```
